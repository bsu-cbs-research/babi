# Capnostream & Mocap Correlation

```python
def zscore(x: np.ndarray) -> np.ndarray:
    return (x - x.mean()) / x.std()

def cosine_similarity(a: np.ndarray, b: np.ndarray):
    a = a / np.linalg.norm(a)
    b = b / np.linalg.norm(b)
    return np.dot(a, b)
```
